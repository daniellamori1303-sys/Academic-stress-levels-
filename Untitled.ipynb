```python
data.columns = data.columns.str.replace('Your Academic Stage','AcademicStage')
data.columns = data.columns.str.replace('Academic pressure from your home','AcademicPressure(home)')
data.columns = data.columns.str.replace('What coping strategy you use as a student?','CopingStrategy')
data.columns = data.columns.str.replace('Do you have any bad habits like smoking, drinking on a daily basis?','Smoking,Drinking,Others?')
data.columns = data.columns.str.replace('What would you rate the academic  competition in your student life','AcademicCompetitionScore')
data.columns = data.columns.str.replace('Rate your academic stress index ','AcademicStressIndex')
data.columns = data.columns.str.replace(' ','')
data.columns
```
